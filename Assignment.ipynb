#Question 1: Student Grade Calculator
Write a Python program that accepts the marks obtained by a student (out of 100) and displays the corresponding grade based on the following criteria:


1.   Marks 90 and above → Grade A
2.   Marks 75 to 89 → Grade B
1.  Marks 60 to 74 → Grade C
2.  Marks 40 to 59 → Grade D
1.   Marks below 40 → Fail

```python
marks=int(input("Enter your marks :"))  #Take marks from user

if marks >100:          #checks that marks are not above 100
  print("Enter correct Marks")
elif marks ==90 or marks>=90:
  print("Grade A")
elif marks >= 75 and marks <=89:
  print("Grade B")
elif marks >= 60 and marks <=74:
  print("Grade C")
elif  marks >=40 and marks <= 59:
  print("Grade D")
else:
  print("Fail")
```

#Question 4: Online Store Delivery Charge Calculator

An online store charges delivery fees based on the order amount and delivery location according to the following conditions:

* If the order amount is ₹1,000 or more, delivery is free.
* If the order amount is less than ₹1,000 and the delivery location is inside the city, the delivery charge is ₹50.
* If the order amount is less than ₹1,000 and the delivery location is outside the city, the delivery charge is ₹100.

Write a Python program that accepts the order amount and delivery location (city or outside) as input and displays the final payable amount, including the delivery charge (if applicable).



```python
amount,location = input("Enter Amount and Location:").split()
amount= float(amount)   #order amount to a float to handle decimal digits

location = location.lower()   # string to lowercase to avoid case-sensitivity issues

if amount >= 1000 :
  delivery_fee =0
elif location == "city":
  delivery_fee = 50
else:
  delivery_fee = 100

order_amount = amount + delivery_fee

print(f"Order Amount : {amount:.2f}")
print(f"Delivery Charges :{delivery_fee:.2f}")
print(f"Total Payable amount :{order_amount:.2f}")
```

#Question 11: Day of the Week

Write a Python program that accepts a number from 1 to 7 and displays the corresponding day of the week using if-elif-else statements.

* 1 → Monday
* 2 → Tuesday
* 3 → Wednesday
* 4 → Thursday
* 5 → Friday
* 6 → Saturday
* 7 → Sunday

If the user enters any number other than 1 to 7, display "Invalid input".

```python
num_value = int(input("Enter number from 0 - 7 only :"))

if num_value == 1:
  print("Day of week 1 is Monday")
elif num_value == 2:
  print("Day of week 2 is Tuesday")
elif num_value == 3:
  print("Day of week 3 is Wednesay")
elif num_value == 4:
  print("Day of week 4 is Thursday")
elif num_value == 5:
  print("Day of week 5 is Friday")
elif num_value == 6:
  print("Day of week 6 is Saturday")
elif num_value == 7:
  print("Day of week 7 is Sunday")
else:
  print("Invalid Input")
```

#Question 12: Simple Calculator
Write a Python program that:

* Accepts two numbers from the user.
* Accepts an operator (+, -, or *) from the user.
* Performs the corresponding arithmetic operation using if-elif-else statements.
* Displays the result.
* If the user enters an operator other than +, -, or *, display "Invalid operation".

```python
x,y = map(int,input("Enter two numbers :").split())

operator = input("Enter an operator :")

if operator == "+":
  result= x + y
  print(f"Addition of two numbers is {result}")
elif operator == "*":
  result= x * y
  print(f"Multiplication of two numbers is {result}")
elif operator == "-":
  result= x - y
  print(f"Subtraction of two numbers is {result}")
else:
  print("Invalid operator")

```
