```python
def book_room(name, **kwargs):
    print("Customer Name:", name)
    print("Booking Details:")

    for key, value in kwargs.items():
        print(key, ":", value)


book_room(
    "Ganesh",
    Room_type="Deluxe",
    Nights=6,
    Breakfast=True,
    City_view=True
)
```

```python
total_tours = 0

def space_tour(visitor_name, destination="Moon"):
    global total_tours

    total_tours += 1
    tour_details = f"{visitor_name} booked a tour to {destination}"

    print(tour_details)

space_tour("Ganesh")
space_tour("Jeevan", "Saturn")
space_tour("Naveen", "Jupiter")
space_tour("Veer","venus")

print("Total tours:", total_tours)
```
