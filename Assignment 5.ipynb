# Create a 5×4 NumPy array of random integers from 10 to 50.

* Find the row with the highest sum.

* Find the average of each column.

* Replace all odd numbers with 0.

```python
import numpy as np
arr = np.random.randint(10,50,size=(5,4))
print("Original Array:")
print(arr,"\n")

row_sum=np.sum(arr,axis=1)
max_index=np.argmax(row_sum)

print("Highest sum row: ",arr[max_index],"\n")
print("Highest Sum:", row_sum[max_index])

col_avg = np.mean(arr, axis=0)
print("\nAvg of each col:",col_avg)

arr[arr%2!=0]=0
print("\nOdd numbers replaced with 0:")
print(arr)
```

# 2. You are given a 4×3 NumPy array:

Rows = quarters

Columns = regions

Values = sales in lakhs



Task:

Calculate average sales per region

Identify the quarter with the lowest overall sales

Print which region had the most stable performance (lowest variance)


```python
sales = np.array([[100,120,150],
                  [130,140,120],
                  [100,160,200],
                  [100,210,180]])

print("sales Data:")
print(sales)

avg_sales = np.mean(sales, axis=0)
print("\nAverage sales per region:")
print("Region 1",avg_sales[0])
print("Region 2",avg_sales[1])
print("Region 3",avg_sales[2])

total_sales = np.sum(sales, axis=1)
min_quarter = np.argmin(total_sales)
print("\nQuarter with lowest overall sales:",min_quarter+1)

var_regions = np.var(sales, axis=0)
most_stable_region = np.argmin(var_regions)
print("\nRegion with most stable performance:",most_stable_region+1)
print("Variance of each region:",var_regions)



```

# 3. You have a list of ticket numbers from 1000 to 1099.



Randomly select 5 winning tickets using np.random.choice().

Ask the user to input their ticket number.

Check if it's a winner.

```python
import numpy as np
ticket = np.arange(1000,1099)

winners = np.random.choice(ticket,size=5,replace=False)

print("Winning Tickets:",winners)

tickets = int(input("Enter your ticket number:"))

if tickets in winners:
  print("Congratulations! You won a prize!")
else:
  print("Sorry!Your not a Winner")
```

# Simulate 30 days of attendance for 10 employees (1 = Present, 0 = Absent).



Generate a 10×30 matrix with values 0 or 1 using randint()

Count how many days each employee was present

Find employees with 90% or more attendance

```python

attendence = np.random.randint(0,2,size=(10,30))
print("Attendance Matrix (1 = Present, 0 = Absent):")
print(attendence)

present_days = np.sum(attendence, axis=1)
print("\nDays Present for each employee:")
for i in range(10):
  print("Employee",i+1,",",present_days[i],"days present")

attendance_percentage = (present_days / 30) * 100

print("\nAttendance Percentage:")
for i in range(10):
    print("Employee", i + 1, ":", attendance_percentage[i], "%")

print("\nEmployees with 90% or more attendance:")
employees = np.where(attendance_percentage >= 90)[0]

if len(employees) > 0:
    for emp in employees:
        print("Employee", emp + 1)
else:
    print("No employees have 90% or more attendance.")
```
