#Question 1: Skip Every Second Character
Task: Write a Python program that:

* Asks the user to enter a word.
* Prints the word by skipping every second character.
* Display only the characters at positions 1st, 3rd, 5th, …

```python
word=input("Enter a word:") # Ask the user to enter words

result=""

for i in range(len(word)):  # lopp through the word using index number

  if i %2==0:
    result +=word[i]
  else:
    pass

print(f"Original word:{word}")
print(f"evenly skipped character :{result}")
```

# Question 2: Digit Sum Until Single Digit
Task: Write a Python program that:

* Takes any positive integer from the user.
* Repeatedly adds all its digits.
* Continues until the result becomes a single digit.
* Prints the final single-digit result.

```python
Num = int(input("Enter any Positive Integer :"))

if Num >=10:

  while Num >=10:
    sumOfDigit = 0
    while Num > 0:
      sumOfDigit += Num % 10
      Num //= 10
    Num = sumOfDigit


  print(sumOfDigit)
else:
  print("Enter Alteast two digit Number..")
```

# Question 5: Cafeteria Billing System
A cafeteria provides the following menu:

Choice	Item	  Price
1. Tea	          ₹10
2. Coffee	        ₹20
3. Juice	        ₹30

Task: Write a Python program that:

Displays the menu.
* Continuously asks the user to enter their choice.
* Allows ordering multiple items.
* If the user types "done", stop taking orders.
* Displays the total bill.
* Shows an appropriate error message for invalid choices.

```python
print("Cafeteria Menu")

print("Item    |Price")
print("Tea     |10")
print("Coffee   |20")
print("Juice   |30")
print("Type 'done' when you are finished ordering.\n")

totalbill=0

while True:
  order=input("Enter item in Menu : ").strip().lower()

  if order == "done":
    break
  if order == "tea":
    totalbill += 10
  elif order == "coffee":
    totalbill += 20
  elif order == "juice":
    totalbill +=30
  else:
    print("Error, Enter valid order..!")
print(f" Your Total Bill is :${totalbill}")



```

# Question 6: Sum of Positive Numbers
Task: Write a Python program that:

* Continuously asks the user to enter numbers.
* Stops when the user enters a negative number.
* Prints the sum of all positive numbers entered.

Hint:

* Use a while loop to keep taking input.
* Use an if statement to check whether the number is negative.
* If negative, exit the loop.
* Otherwise, add the number to a sum variable and continue.



```python
sum=0
while True:
  pos_num = int(input("Enter Positive Number :"))

  if pos_num < 0:
    break

  if pos_num >0:
    sum += pos_num
print(f"Sum of Positive Integers is : {sum}")
```
