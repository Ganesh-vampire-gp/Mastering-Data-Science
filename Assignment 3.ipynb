# Question 1: Meal Ordering System (Default Arguments)
A restaurant allows customers to order a meal with an optional drink.

Task: Write a Python function order_meal() that takes:

* main_dish (required argument)
* drink (default argument with the default value "Water")
* The function should print the customer's meal order.

```python
def order_meal(main_dish,drink="water"):

  print("----Customer Order Details----\n")

  print("\tMain Dish :",main_dish)
  print("\tDrink     :",drink)
  print("Thank You For Ordering...:)\n")

order_meal("Briyani")
order_meal("Omlatte","Coke")
```

# Question 2: Return Double the Value
Task: Write a function that:

* Takes a number as input.
* Returns double its value.
* Call the function with values 4, 6, and 10.
* Print the returned results.

```python
def double_num(num):
  print(f"Double of {num} is :{num*2}")
double_num(6)
double_num(4)
double_num(10)
```

# Question 3: Using a Global Variable Inside a Function
Create a global variable:

* animal = "tiger"



Task:

* Write a function that prints:
* " My favorite animal is tiger " using the global variable.



```python
animal = "Tiger"
def Golb_animal():
  global animal
  print(f"My favorite animal is {animal}")
Golb_animal()
```

# Question 5: Using the global Keyword
Task:

Create a variable:

place = "park"



* Inside a function, use the global keyword to make it a global variable.
* Print its value inside the function.
* Call the function.
* Print the value outside the function.

```python
place ="park"

def Place():
  global place
  print(place)
Place()
print(place)
```

# Question 6: Modifying a Global Variable
Create a global variable:

* weather = "sunny"



Task:

* Print the value of weather.

Inside a function, use the global keyword to change its value to:
"rainy"



Call the function.
* Print the updated value of weather.



```python
weather = "sunny"
print(weather)

def cloud():
  global weather
  weather = "rainy"
cloud()
print(weather)
```
